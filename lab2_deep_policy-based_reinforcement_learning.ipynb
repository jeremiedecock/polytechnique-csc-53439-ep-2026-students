## Setup the Gymnasium environment

As in the previous lab, we will try to solve the CartPole-v1 environment (see: https://gymnasium.farama.org/environments/classic_control/cart_pole/), which offers a continuous state space and a discrete action space.

**Reminder**: the CartPole task consists of keeping a pole upright by moving the cart to which it is attached via a joint.
No friction is considered.
An episode ends when the pole is more than 12 degrees from vertical, when the cart moves more than 2.4 units from the center, or after 500 steps.
A reward of +1 is given for every step taken, so the return of an episode is at most 500.
The task is considered solved when the average return over 100 consecutive episodes is at least 475.
The state is given by $(x,\dot{x},\phi,\dot{\phi})$ where $x$ is the position of the cart and $\phi$ is the angle between the pole and the vertical.
There are only two possible actions: $a \in \{0, 1\}$ where $a = 0$ means "push the cart to the LEFT" and $a = 1$ means "push the cart to the RIGHT".

````python
env = gym.make("CartPole-v1", render_mode="rgb_array")

cartpole_observation_space = cast(gym.spaces.Box, env.observation_space)
cartpole_action_space = cast(gym.spaces.Discrete, env.action_space)

cartpole_observation_dim:int = cartpole_observation_space.shape[0]    # Number of *dimensions* in the observation space (i.e. number of *elements* in the observation vector)
cartpole_num_actions:int = cast(int, cartpole_action_space.n.item())  # Number of possible actions

print(f"State space is: {cartpole_observation_space}")
print(f"Action space is: {cartpole_action_space}")
print("Actions are: {" + ", ".join([str(a) for a in range(cartpole_num_actions)]) + "}")

env.close()
````

## Part 1: Monte Carlo Policy Gradient (REINFORCE)

### The Policy Gradient theorem

We will solve the CartPole environment using a policy gradient method which directly searches, within a family of parameterized policies $\pi_{\boldsymbol{\theta}}$, for an optimal policy.

This method performs gradient ascent in the policy space so that the total return is maximized.
We will restrict our work to episodic tasks, *i.e.* tasks that have a starting state and last for at most a finite number of steps $T$, called the horizon ($T = 500$ for CartPole-v1).
When an episode terminates before the horizon (e.g. when the pole falls), we consider that the agent remains in an absorbing terminal state where all subsequent rewards are zero, so that all the sums below can be written up to $T-1$.

More formally, we define an optimization criterion that we want to maximize:

$$J(\boldsymbol{\theta}) = \mathbb{E}_{\pi_{\boldsymbol{\theta}}}\left[\sum_{t=0}^{T-1} r(s_t,a_t)\right],$$

where $\mathbb{E}_{\pi_{\boldsymbol{\theta}}}$ denotes the expectation over the trajectories generated by following $\pi_{\boldsymbol{\theta}}$ (*i.e.* $a_t \sim \pi_{\boldsymbol{\theta}}(\cdot|s_t)$ at each time step $t$) and $T$ is the horizon of the episode.
In other words, we want to maximize the expected value of the initial state: $J(\boldsymbol{\theta}) = \mathbb{E}_{s_0}\left[V^{\pi_{\boldsymbol{\theta}}}(s_0)\right]$ (in CartPole, the initial state $s_0$ is randomly drawn at the beginning of each episode).
The policy gradient theorem tells us that:

$$
\nabla_{\boldsymbol{\theta}} J(\boldsymbol{\theta}) = \mathbb{E}_{\pi_{\boldsymbol{\theta}}} \left[\sum_{t=0}^{T-1} \nabla_{\boldsymbol{\theta}} \ln \pi_{\boldsymbol{\theta}} (a_t|s_t) ~ Q^{\pi_{\boldsymbol{\theta}}}(s_t,a_t) \right],
$$

where $Q^{\pi_{\boldsymbol{\theta}}}(s_t,a_t)$ is the expected return obtained from time step $t$ onwards, after taking action $a_t$ in state $s_t$ and then following $\pi_{\boldsymbol{\theta}}$:

$$Q^{\pi_{\boldsymbol{\theta}}}(s_t,a_t) = \mathbb{E}_{\pi_{\boldsymbol{\theta}}} \left[\sum_{t'=t}^{T-1} r(s_{t'},a_{t'}) ~\middle|~ s_t, a_t\right].$$

Note that with a finite horizon, the $Q$-function also depends on the time step $t$ (the number of remaining steps decreases as $t$ increases); this dependency is omitted in the notation for readability.

The policy gradient theorem is particularly effective because it allows gradient computation without needing to model the system's dynamics, as long as the $Q$-function (or an estimate of it) is available. By simply applying the policy and observing one-step transitions, sufficient information can be gathered. Using stochastic gradient ascent and substituting $Q^{\pi_{\boldsymbol{\theta}}}(s_t,a_t)$ with a Monte Carlo estimate $G_t = \sum_{t'=t}^{T-1} r(s_{t'},a_{t'})$ for a single trajectory, we obtain the REINFORCE algorithm.

The REINFORCE algorithm, introduced by Williams in 1992, is a Monte Carlo policy gradient method. It updates the policy parameters along a stochastic estimate of the gradient $\nabla_{\boldsymbol{\theta}} J(\boldsymbol{\theta})$, in which the full-episode returns $G_t$ are used as unbiased estimates of $Q^{\pi_{\boldsymbol{\theta}}}(s_t,a_t)$. Each step involves generating an episode using the current policy, computing the gradient estimate, and updating the policy parameters. This algorithm is simple and model-free (it does not require any knowledge of the environment dynamics), but its gradient estimates have a high variance, which can make learning slow and unstable. Exercise 2 introduces a *baseline* to reduce this variance.

For further reading and a deeper understanding, refer to Williams' seminal paper (https://link.springer.com/article/10.1007/BF00992696) and the comprehensive text on reinforcement learning by Richard S. Sutton and Andrew G. Barto: "Reinforcement Learning: An Introduction", chap.13 (http://incompleteideas.net/book/RLbook2020.pdf).

Here is the REINFORCE algorithm.

#### Monte Carlo policy gradient (REINFORCE)

<b>REQUIRE</b> <br>
$\quad$ A differentiable policy $\pi_{\boldsymbol{\theta}}$ <br>
$\quad$ A learning rate $\alpha \in \mathbb{R}^+$ <br>
<b>INITIALIZATION</b> <br>
$\quad$ Initialize parameters $\boldsymbol{\theta} \in \mathbb{R}^d$ <br>
<br>
<b>FOR EACH</b> episode <br>
$\quad$ Generate full trace $\tau = ( \boldsymbol{s}_0, \boldsymbol{a}_0, r_0, \boldsymbol{s}_1, \boldsymbol{a}_1, \dots, r_{T-1}, \boldsymbol{s}_T )$ following $\pi_{\boldsymbol{\theta}}$ <br>
$\quad$ <b>FOR</b> $~ t=0,\dots,T-1$ <br>
$\quad\quad$ $G_t \leftarrow \sum_{k=t}^{T-1} r_k$ <br>
$\quad$ $\boldsymbol{\theta} \leftarrow \boldsymbol{\theta} + \alpha ~ \underbrace{\sum_{t=0}^{T-1} G_t ~ \nabla_{\boldsymbol{\theta}} \ln \pi_{\boldsymbol{\theta}}(\boldsymbol{a}_t|\boldsymbol{s}_t)}_{\text{estimate of } \nabla_{\boldsymbol{\theta}} J(\boldsymbol{\theta})}$ <br>
<br>
<b>RETURN</b> $\boldsymbol{\theta}$

### Exercise 2: REINFORCE with Baseline [reminder]

In the basic REINFORCE algorithm, the policy parameters are updated in proportion to the product of the gradient of the policy log-probability and the cumulative reward (return) from a state-action pair. However, this approach can lead to high variance in policy updates, making learning slower.

In this exercise, a *baseline* is introduced to reduce this variance. It is a value subtracted from the cumulative reward when calculating the policy gradient. The key property of the baseline is that, as long as it does not depend on the action (it can depend on the state), it does not affect the expected value of the gradient estimate: it does not bias the learning process, but it can reduce the variance of the updates.

The baseline can be thought of as a reference point or an "average" expectation of reward. By comparing the actual rewards to this baseline, we can determine whether the outcomes of certain actions are better or worse than this "average" performance.

A common choice for the baseline is an estimate $\hat{V}_{\boldsymbol{\omega}}$ of the value function $V^{\pi_{\boldsymbol{\theta}}}$ of the current policy. By using the value function as a baseline, the algorithm adjusts the policy towards actions that perform better than the average.

To incorporate the baseline into REINFORCE, you modify the update rule. Instead of using the return $G_t$ directly, you subtract the baseline value $\hat{V}_{\boldsymbol{\omega}}(s_t)$ from $G_t$ in the policy gradient estimate. The difference $\hat{A}_t = G_t - \hat{V}_{\boldsymbol{\omega}}(s_t)$ is an estimate of the *advantage* $A^{\pi_{\boldsymbol{\theta}}}(s_t, a_t) = Q^{\pi_{\boldsymbol{\theta}}}(s_t, a_t) - V^{\pi_{\boldsymbol{\theta}}}(s_t)$ of action $a_t$, *i.e.* of how much better $a_t$ is than the average action of the policy in $s_t$. All the following algorithms of this lab weight $\nabla_{\boldsymbol{\theta}} \ln \pi_{\boldsymbol{\theta}}(a_t|s_t)$ by such an advantage estimate: they only differ in the way $\hat{A}_t$ is computed.

By centering the returns around a baseline, the variance of the policy gradient estimates is reduced. This leads to more stable and efficient learning, as the updates are less noisy and more focused on improving relative to the average performance.

## Part 2: Bootstrapping (Actor-Critic)

### From Monte Carlo to bootstrapping

In Part 1, the policy is updated with the Monte Carlo return $G_t$, an unbiased estimate of $Q^{\pi_{\boldsymbol{\theta}}}(s_t, a_t)$.
This estimate has two drawbacks: it is only known at the end of the episode, and it has a high variance because it depends on all the random actions and transitions from time step $t$ until the end of the episode.

**Actor-Critic** methods replace $G_t$ with an estimate given by a learned value function, the *critic*, which is trained by **bootstrapping**: as for the Q-networks of the previous lab, its target is built from its own estimate of the value of the next state (*e.g.* $r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1})$).
An Actor-Critic algorithm thus learns two functions:
- **the actor** $\pi_{\boldsymbol{\theta}}$: the policy, which selects the actions (we reuse the `PolicyNetwork` of Exercise 1);
- **the critic** ($\hat{V}_{\boldsymbol{\omega}}$ or $\hat{Q}_{\boldsymbol{\omega}}$): an estimate of the value function of the current policy, which evaluates the actions selected by the actor.

Bootstrapping has two benefits: the networks can be updated at every time step, without waiting for the end of the episode, and the target only depends on one random action and transition, which reduces the variance.
The price to pay is a **bias**: the target relies on the critic, which is inaccurate, especially at the beginning of the training.
This **bias-variance trade-off** is the common thread of this part.

Note that the value network of Exercise 2 is not a critic: it is trained on Monte Carlo returns and only used as a baseline.
Following Sutton and Barto ("Reinforcement Learning: An Introduction", §13.5), a value function is called a critic only when it is used for bootstrapping (some authors, *e.g.* David Silver in his lectures, use the word more loosely).

The exercises of this part are summarized below. In all of them, the policy is updated in the direction of $\hat{A}_t \nabla_{\boldsymbol{\theta}} \ln \pi_{\boldsymbol{\theta}}(a_t|s_t)$: what changes is the function learned by the critic and the way the advantage is estimated.

|                                               | Value network                                    | Target of the value network                                                                                                     | Advantage estimate $\hat{A}_t$                                                                                                      |
|-----------------------------------------------|--------------------------------------------------|---------------------------------------------------------------------------------------------------------------------------------|-------------------------------------------------------------------------------------------------------------------------------------|
| Exercise 2 (reminder): REINFORCE with baseline | $\hat{V}_{\boldsymbol{\omega}}$ (baseline)       | Monte Carlo return $G_t$                                                                                                        | $G_t - \hat{V}_{\boldsymbol{\omega}}(s_t)$                                                                                          |
| Exercise 3: TD Actor-Critic                   | $\hat{V}_{\boldsymbol{\omega}}$ (critic)         | $r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1})$                                                                           | $\delta_t = r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1}) - \hat{V}_{\boldsymbol{\omega}}(s_t)$                               |
| Exercise 4: Q Actor-Critic                    | $\hat{Q}_{\boldsymbol{\omega}}$ (critic)         | $r_t + \gamma \hat{Q}_{\boldsymbol{\omega}}(s_{t+1}, a')$, with $a' \sim \pi_{\boldsymbol{\theta}}(\cdot \mid s_{t+1})$          | $\hat{Q}_{\boldsymbol{\omega}}(s_t, a_t) - \sum_{a} \pi_{\boldsymbol{\theta}}(a \mid s_t) ~ \hat{Q}_{\boldsymbol{\omega}}(s_t, a)$ |
| Exercise 5: Generalized Advantage Estimation  | $\hat{V}_{\boldsymbol{\omega}}$ (critic)         | $\hat{A}_t + \hat{V}_{\boldsymbol{\omega}}(s_t)$ ($\lambda$-return)                                                             | $\sum_{k \geq 0} (\gamma \lambda)^k ~ \delta_{t+k}$                                                                                 |

### Why do we need a discount factor now?

In Part 1, we used undiscounted returns ($\gamma = 1$).
This was not a problem: the episodes last at most $T = 500$ steps, so the returns are bounded, and the value network of Exercise 2 was only a baseline, so it did not matter that it could only approximate a return that also depends on the number of remaining steps (see Exercise 2).

From now on, we use a discount factor $\gamma = 0.99$: the value of a state is the expected *discounted* return $V^{\pi_{\boldsymbol{\theta}}}(s) = \mathbb{E}_{\pi_{\boldsymbol{\theta}}}\left[\sum_{k=0}^{\infty} \gamma^k r_{t+k} ~\middle|~ s_t = s\right]$.
The discount factor is needed as soon as the critic bootstraps:

1. **Truncated episodes must be bootstrapped.** When CartPole-v1 stops an episode after 500 steps, the last state is not a terminal state: the pole is still up and the agent could have continued. As in the previous lab, its value must therefore be bootstrapped (use `terminated`, not `terminated or truncated`, to detect terminal states). The critic thus estimates the value of a task that may last forever, which only depends on the state (the time step is not part of the state). With $\gamma = 1$, this value would be infinite for a good policy, which receives $+1$ at every time step: the TD targets would grow without bound and the critic would diverge. With $\gamma < 1$, the values are bounded: $V^{\pi_{\boldsymbol{\theta}}}(s) \leq \frac{r_{\max}}{1 - \gamma} = 100$ for $\gamma = 0.99$.
2. **Discounting stabilizes bootstrapping and reduces the variance.** At each bootstrapping step, the error of the critic on the next state is multiplied by $\gamma$, so errors are attenuated instead of accumulating (in the tabular case, the Bellman operator is a $\gamma$-contraction). Moreover, the rewards obtained more than about $\frac{1}{1 - \gamma} = 100$ steps later (the *effective horizon*) have little weight, which reduces the variance of the targets.

Discounting does not change the goal in CartPole: since all rewards are $+1$, the discounted return of an episode of length $L$ is $\frac{1 - \gamma^L}{1 - \gamma}$, which increases with $L$.
The best policies are still those that keep the pole up as long as possible, and we still evaluate the policies with the undiscounted return (`avg_return_on_multiple_episodes`).

**Note**: strictly speaking, the gradient of the discounted objective contains an additional factor $\gamma^t$ in front of each term $\nabla_{\boldsymbol{\theta}} \ln \pi_{\boldsymbol{\theta}}(a_t|s_t)$ (see the "one-step actor-critic" pseudo-code in Sutton and Barto, §13.5). As in most implementations (A2C, PPO, etc.), we omit it: the resulting update is not exactly the gradient of the discounted objective, but it works well in practice.

### Exercise 3: TD Actor-Critic

Our first actor-critic algorithm is obtained from REINFORCE with baseline (Exercise 2) with a single change: the Monte Carlo return $G_t$ is replaced with the bootstrapped (TD) target $r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1})$.
The value network $\hat{V}_{\boldsymbol{\omega}}$ thus becomes a critic, and the advantage estimate $G_t - \hat{V}_{\boldsymbol{\omega}}(s_t)$ of Exercise 2 becomes the **TD error** of the critic:

$$\delta_t = r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1}) - \hat{V}_{\boldsymbol{\omega}}(s_t).$$

**Why is the TD error an estimate of the advantage?** If the critic is exact ($\hat{V}_{\boldsymbol{\omega}} = V^{\pi_{\boldsymbol{\theta}}}$), the expected value of the TD error, given $s_t$ and $a_t$, is the advantage of $a_t$:

$$\mathbb{E}\left[\delta_t ~\middle|~ s_t, a_t\right] = \underbrace{\mathbb{E}\left[r_t + \gamma V^{\pi_{\boldsymbol{\theta}}}(s_{t+1}) ~\middle|~ s_t, a_t\right]}_{Q^{\pi_{\boldsymbol{\theta}}}(s_t, a_t)} - V^{\pi_{\boldsymbol{\theta}}}(s_t) = A^{\pi_{\boldsymbol{\theta}}}(s_t, a_t).$$

The same quantity $\delta_t$ is therefore used twice: to train the critic (by minimizing $\delta_t^2$) and to update the policy (as an advantage estimate $\hat{A}_t = \delta_t$).
Unlike $G_t - \hat{V}_{\boldsymbol{\omega}}(s_t)$, it only depends on one random action and one random transition, so it has a much lower variance; but it is biased as long as the critic is inaccurate.

This algorithm is called *TD Actor-Critic* in David Silver's lectures and *one-step actor-critic* in Sutton and Barto (§13.5). It is also the one-step version of A2C (Mnih et al., 2016, https://arxiv.org/abs/1602.01783).

The main differences with REINFORCE with baseline (Exercise 2) are summarized below:

|                                                                                       | REINFORCE with baseline (Exercise 2)               | TD Actor-Critic (Exercise 3)                                                                               |
|---------------------------------------------------------------------------------------|----------------------------------------------------|------------------------------------------------------------------------------------------------------------|
| Target of $\hat{V}_{\boldsymbol{\omega}}(s_t)$                                        | Monte Carlo return $G_t$                           | Bootstrapped (TD) target $r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1})$                             |
| Weight of $\nabla_{\boldsymbol{\theta}} \ln \pi_{\boldsymbol{\theta}}(a_t \mid s_t)$ | $G_t - \hat{V}_{\boldsymbol{\omega}}(s_t)$         | $\delta_t = r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1}) - \hat{V}_{\boldsymbol{\omega}}(s_t)$      |
| Updates                                                                               | Once per episode, at the end of the episode        | At every time step (online)                                                                                |
| Discount factor                                                                       | $\gamma = 1$                                       | $\gamma = 0.99$                                                                                            |

#### TD Actor-Critic

<b>REQUIRE</b> <br>
$\quad$ A differentiable policy $\pi_{\boldsymbol{\theta}}$ <br>
$\quad$ A differentiable state-value function $\hat{V}_{\boldsymbol{\omega}}(\boldsymbol{s})$ <br>
$\quad$ A learning rate $\alpha_1 \in \mathbb{R}^+$ for the policy (actor) <br>
$\quad$ A learning rate $\alpha_2 \in \mathbb{R}^+$ for the state-value function (critic) <br>
$\quad$ A discount factor $\gamma \in [0, 1)$ <br>
<b>INITIALIZATION</b> <br>
$\quad$ Initialize parameters $\boldsymbol{\theta} \in \mathbb{R}^d$ <br>
$\quad$ Initialize parameters $\boldsymbol{\omega} \in \mathbb{R}^{d'}$ <br>
<br>
<b>FOR EACH</b> episode <br>
$\quad$ $\boldsymbol{s}_0 \leftarrow$ initial state <br>
$\quad$ <b>FOR</b> $~ t=0,1,\dots$ until the episode ends <br>
$\quad\quad$ Sample $\boldsymbol{a}_t \sim \pi_{\boldsymbol{\theta}}(\cdot|\boldsymbol{s}_t)$, take action $\boldsymbol{a}_t$ and observe $r_t$ and $\boldsymbol{s}_{t+1}$ <br>
$\quad\quad$ $\delta_t \leftarrow r_t + \gamma ~ \hat{V}_{\boldsymbol{\omega}}(\boldsymbol{s}_{t+1}) - \hat{V}_{\boldsymbol{\omega}}(\boldsymbol{s}_t)$ $\quad$ (with $\hat{V}_{\boldsymbol{\omega}}(\boldsymbol{s}_{t+1}) = 0$ if $\boldsymbol{s}_{t+1}$ is a terminal state) <br>
$\quad\quad$ $\boldsymbol{\theta} \leftarrow \boldsymbol{\theta} + \alpha_1 ~ \delta_t ~ \nabla_{\boldsymbol{\theta}} \ln \pi_{\boldsymbol{\theta}}(\boldsymbol{a}_t|\boldsymbol{s}_t)$ <br>
$\quad\quad$ $\boldsymbol{\omega} \leftarrow \boldsymbol{\omega} + \alpha_2 ~ \delta_t ~ \nabla_{\boldsymbol{\omega}} \hat{V}_{\boldsymbol{\omega}}(\boldsymbol{s}_t)$ <br>
<br>
<b>RETURN</b> $\boldsymbol{\theta}$

#### Implement the train function

**Task 3.1**: Implement the `train_td_actor_critic_discrete` function, used to train a policy network (actor) and a value network (critic) using the TD Actor-Critic algorithm in a given environment.

No new network is needed: as in Exercise 2, the function first initializes a `PolicyNetwork` and a `ValueNetwork`, along with their respective Adam optimizers.
The main difference is that the networks are updated **at every time step** of the episode, and not once at the end of the episode: `sample_one_episode` is therefore replaced by a loop over the time steps of the episode.
At each time step $t$, once the action $a_t$ has been sampled and executed, the function:
1. computes the TD target $r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1})$ (bootstrapping). As in the previous lab, the gradient must not flow through the target: compute it in a [torch.no_grad()](https://docs.pytorch.org/docs/stable/generated/torch.no_grad.html) block;

### Exercise 4: Q Actor-Critic (QAC)

In Exercise 3, the critic estimates the state-value function $V^{\pi_{\boldsymbol{\theta}}}$.
In this exercise, the critic $\hat{Q}_{\boldsymbol{\omega}}$ estimates the **action-value function** $Q^{\pi_{\boldsymbol{\theta}}}$, like the Q-networks of the previous lab.
This is the most direct use of the policy gradient theorem: REINFORCE replaces $Q^{\pi_{\boldsymbol{\theta}}}(s_t, a_t)$ with $G_t$, whereas QAC replaces it with $\hat{Q}_{\boldsymbol{\omega}}(s_t, a_t)$.

A Q critic can evaluate any action in a given state, not only the action selected by the actor.
This is the starting point of actor-critic algorithms such as DDPG, TD3 and SAC (SAC is used in the bonus section), which can handle continuous actions and reuse past transitions stored in a replay buffer (off-policy learning).
TD Actor-Critic (Exercise 3) is instead the ancestor of A2C and PPO.

As in Exercises 2 and 3, we subtract a baseline to reduce the variance of the policy updates. This baseline is directly computed from the critic, so no additional network is needed:

$$\hat{V}(s) = \sum_{a} \pi_{\boldsymbol{\theta}}(a|s) ~ \hat{Q}_{\boldsymbol{\omega}}(s, a),$$

which is an estimate of $V^{\pi_{\boldsymbol{\theta}}}(s) = \mathbb{E}_{a \sim \pi_{\boldsymbol{\theta}}(\cdot|s)}\left[Q^{\pi_{\boldsymbol{\theta}}}(s, a)\right]$.
The policy update then uses $\hat{A}_t = \hat{Q}_{\boldsymbol{\omega}}(s_t, a_t) - \hat{V}(s_t)$, another estimate of the advantage of $a_t$: it is positive if $a_t$ is better than the average action of the policy in $s_t$, and negative otherwise.
This baseline is essential here: without it (*i.e.* with the "vanilla" QAC update $\boldsymbol{\theta} \leftarrow \boldsymbol{\theta} + \alpha_1 \hat{Q}_{\boldsymbol{\omega}}(s_t, a_t) \nabla_{\boldsymbol{\theta}} \ln \pi_{\boldsymbol{\theta}}(a_t|s_t)$), the agent barely learns anything on CartPole within the training budget used below.

The main differences with TD Actor-Critic (Exercise 3) are summarized below:

|                                                                                       | TD Actor-Critic (Exercise 3)                                                                           | QAC (Exercise 4)                                                                                                                  |
|---------------------------------------------------------------------------------------|--------------------------------------------------------------------------------------------------------|-----------------------------------------------------------------------------------------------------------------------------------|
| Critic                                                                                | $\hat{V}_{\boldsymbol{\omega}}(s)$ (state value)                                                       | $\hat{Q}_{\boldsymbol{\omega}}(s, a)$ (action value)                                                                              |
| Target of the critic                                                                  | $r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1})$                                                  | $r_t + \gamma \hat{Q}_{\boldsymbol{\omega}}(s_{t+1}, a')$, with $a' \sim \pi_{\boldsymbol{\theta}}(\cdot \mid s_{t+1})$            |
| Weight of $\nabla_{\boldsymbol{\theta}} \ln \pi_{\boldsymbol{\theta}}(a_t \mid s_t)$ | $\delta_t = r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1}) - \hat{V}_{\boldsymbol{\omega}}(s_t)$  | $\hat{Q}_{\boldsymbol{\omega}}(s_t, a_t) - \hat{V}(s_t)$                                                                          |

#### Q Actor-Critic (QAC)

<b>REQUIRE</b> <br>
$\quad$ A differentiable policy $\pi_{\boldsymbol{\theta}}$ <br>
$\quad$ A differentiable action-value function $\hat{Q}_{\boldsymbol{\omega}}(\boldsymbol{s}, \boldsymbol{a})$ <br>
$\quad$ A learning rate $\alpha_1 \in \mathbb{R}^+$ for the policy (actor) <br>
$\quad$ A learning rate $\alpha_2 \in \mathbb{R}^+$ for the action-value function (critic) <br>
$\quad$ A discount factor $\gamma \in [0, 1)$ <br>
<b>INITIALIZATION</b> <br>
$\quad$ Initialize parameters $\boldsymbol{\theta} \in \mathbb{R}^d$ <br>
$\quad$ Initialize parameters $\boldsymbol{\omega} \in \mathbb{R}^{d'}$ <br>
<br>
<b>FOR EACH</b> episode <br>
$\quad$ $\boldsymbol{s}_0 \leftarrow$ initial state <br>
$\quad$ <b>FOR</b> $~ t=0,1,\dots$ until the episode ends <br>
$\quad\quad$ Sample $\boldsymbol{a}_t \sim \pi_{\boldsymbol{\theta}}(\cdot|\boldsymbol{s}_t)$, take action $\boldsymbol{a}_t$ and observe $r_t$ and $\boldsymbol{s}_{t+1}$ <br>
$\quad\quad$ Sample $\boldsymbol{a}' \sim \pi_{\boldsymbol{\theta}}(\cdot|\boldsymbol{s}_{t+1})$ $\quad$ (only used to compute the TD target) <br>
$\quad\quad$ $\delta_t \leftarrow r_t + \gamma ~ \hat{Q}_{\boldsymbol{\omega}}(\boldsymbol{s}_{t+1}, \boldsymbol{a}') - \hat{Q}_{\boldsymbol{\omega}}(\boldsymbol{s}_t, \boldsymbol{a}_t)$ $\quad$ (with $\hat{Q}_{\boldsymbol{\omega}}(\boldsymbol{s}_{t+1}, \boldsymbol{a}') = 0$ if $\boldsymbol{s}_{t+1}$ is a terminal state) <br>
$\quad\quad$ $\boldsymbol{\theta} \leftarrow \boldsymbol{\theta} + \alpha_1 \left( \hat{Q}_{\boldsymbol{\omega}}(\boldsymbol{s}_t, \boldsymbol{a}_t) - \hat{V}(\boldsymbol{s}_t) \right) \nabla_{\boldsymbol{\theta}} \ln \pi_{\boldsymbol{\theta}}(\boldsymbol{a}_t|\boldsymbol{s}_t)$ <br>
$\quad\quad$ $\boldsymbol{\omega} \leftarrow \boldsymbol{\omega} + \alpha_2 ~ \delta_t ~ \nabla_{\boldsymbol{\omega}} \hat{Q}_{\boldsymbol{\omega}}(\boldsymbol{s}_t, \boldsymbol{a}_t)$ <br>
<br>
<b>RETURN</b> $\boldsymbol{\theta}$

where $\hat{V}(\boldsymbol{s}) = \sum_{\boldsymbol{a}} \pi_{\boldsymbol{\theta}}(\boldsymbol{a}|\boldsymbol{s}) ~ \hat{Q}_{\boldsymbol{\omega}}(\boldsymbol{s}, \boldsymbol{a})$.

Note that the TD error $\delta_t$ is only used to train the critic. Unlike the TD error of $\hat{V}_{\boldsymbol{\omega}}$ in Exercise 3, the TD error of $\hat{Q}_{\boldsymbol{\omega}}$ is *not* an estimate of the advantage of $\boldsymbol{a}_t$: when the critic is exact, its expected value given $\boldsymbol{s}_t$ and $\boldsymbol{a}_t$ is zero, since $\mathbb{E}\left[r_t + \gamma Q^{\pi_{\boldsymbol{\theta}}}(\boldsymbol{s}_{t+1}, \boldsymbol{a}') ~\middle|~ \boldsymbol{s}_t, \boldsymbol{a}_t\right] = Q^{\pi_{\boldsymbol{\theta}}}(\boldsymbol{s}_t, \boldsymbol{a}_t)$. It must therefore not be used to update the policy.

**Task 4.2**: Implement the `train_qac_discrete` function, used to train a policy network (actor) and a Q-value network (critic) using the QAC algorithm in a given environment.

The function has the same structure as `train_td_actor_critic_discrete` (Exercise 3): it initializes a policy network and a Q-value network, along with their respective Adam optimizers, and it updates both networks **at every time step** of the episode.
At each time step $t$, once the action $a_t$ has been sampled and executed, the function:
1. samples an action $a' \sim \pi_{\boldsymbol{\theta}}(\cdot|s_{t+1})$ with `sample_discrete_action` and computes the TD target $r_t + \gamma \hat{Q}_{\boldsymbol{\omega}}(s_{t+1}, a')$ (bootstrapping). As in Exercise 3, the gradient must not flow through the target: compute it in a [torch.no_grad()](https://docs.pytorch.org/docs/stable/generated/torch.no_grad.html) block;

As in Exercises 2 and 3, the advantage estimate $\hat{Q}_{\boldsymbol{\omega}}(s_t, a_t) - \hat{V}(s_t)$ must be detached from the computation graph, so that the policy loss does not update the critic;

- `policy_nn(state_tensor)` returns the probabilities $\pi_{\boldsymbol{\theta}}(\cdot|s)$ and `q_nn(state_tensor)` returns the Q-values $\hat{Q}_{\boldsymbol{\omega}}(s, \cdot)$, both as tensors of shape (1, n_actions): $\hat{V}(s)$ is simply `(policy_nn(state_tensor) * q_nn(state_tensor)).sum()`.

### Bonus exercise: n-step returns and Generalized Advantage Estimation (GAE)

Exercises 2 and 3 are the two extremes of the bias-variance trade-off:
- REINFORCE with baseline uses the Monte Carlo advantage estimate $G_t - \hat{V}_{\boldsymbol{\omega}}(s_t)$: it is unbiased, but it has a high variance, since $G_t$ sums all the random rewards until the end of the episode;
- TD Actor-Critic uses the TD error $\delta_t = r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1}) - \hat{V}_{\boldsymbol{\omega}}(s_t)$: it has a low variance, since it only depends on one reward, but it is biased as long as the critic is inaccurate.

**n-step returns.** Between these two extremes, we can sum the actual rewards of the next $n$ time steps and only bootstrap after them:

$$G_t^{(n)} = r_t + \gamma r_{t+1} + \dots + \gamma^{n-1} r_{t+n-1} + \gamma^n \hat{V}_{\boldsymbol{\omega}}(s_{t+n}).$$

With $n = 1$, $G_t^{(1)} = r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1})$ is the TD target of Exercise 3; when $t + n$ reaches the end of the episode, $G_t^{(n)}$ is the Monte Carlo return $G_t$ of Exercise 2.
The corresponding advantage estimate is a discounted sum of TD errors (the intermediate values $\hat{V}_{\boldsymbol{\omega}}(s_{t+1}), \dots, \hat{V}_{\boldsymbol{\omega}}(s_{t+n-1})$ cancel out):

$$\hat{A}_t^{(n)} = G_t^{(n)} - \hat{V}_{\boldsymbol{\omega}}(s_t) = \sum_{k=0}^{n-1} \gamma^k \delta_{t+k}.$$

The larger $n$, the smaller the bias (the critic only intervenes after $n$ steps, with a weight $\gamma^n$) and the larger the variance (more random rewards are summed). A2C (Mnih et al., 2016) typically uses $n = 5$.

**Generalized Advantage Estimation.** Rather than choosing a single value of $n$, GAE (Schulman et al., 2016, https://arxiv.org/abs/1506.02438) averages all the n-step advantage estimates with exponentially decreasing weights $(1 - \lambda) \lambda^{n-1}$, where $\lambda \in [0, 1]$:

$$\hat{A}_t = (1 - \lambda) \sum_{n=1}^{\infty} \lambda^{n-1} \hat{A}_t^{(n)} = \sum_{k=0}^{\infty} (\gamma \lambda)^k ~ \delta_{t+k},$$

with $\delta_{t+k} = 0$ after the end of the episode. The parameter $\lambda$ controls the bias-variance trade-off:
- $\lambda = 0$: $\hat{A}_t = \delta_t$, the advantage estimate of TD Actor-Critic (Exercise 3);
- $\lambda = 1$: $\hat{A}_t = \sum_{k} \gamma^k \delta_{t+k} = G_t - \hat{V}_{\boldsymbol{\omega}}(s_t)$, the advantage estimate of REINFORCE with baseline (Exercise 2), here with a discounted return (and with the bootstrapped value $\gamma^{T-t} \hat{V}_{\boldsymbol{\omega}}(s_T)$ added to $G_t$ when the episode is truncated);
- intermediate values (typically $\lambda = 0.95$, the default value in PPO) usually give a better trade-off.

GAE is the advantage estimator used by PPO (see the bonus section). It is the modern counterpart of the *TD($\lambda$) Actor-Critic* of David Silver's lectures, which achieves the same trade-off online with eligibility traces.

In practice, GAE is computed backward, from the end of the episode, with the recursive formula

$$\hat{A}_t = \delta_t + \gamma \lambda ~ \hat{A}_{t+1}, \quad \text{with } \hat{A}_T = 0.$$

Since $\hat{A}_t$ depends on the future TD errors, the networks are updated once per episode, at the end of the episode, as in Exercise 2.
The critic is trained to predict the *$\lambda$-return* $\hat{A}_t + \hat{V}_{\boldsymbol{\omega}}(s_t)$, which is the TD target $r_t + \gamma \hat{V}_{\boldsymbol{\omega}}(s_{t+1})$ for $\lambda = 0$ and the return $G_t$ for $\lambda = 1$.
Note that with $\lambda = 0$, this algorithm is not exactly the TD Actor-Critic of Exercise 3: the advantage estimates are the same, but the networks are updated once per episode instead of at every time step.

**Task 5.3**: Implement the `train_gae_discrete` function, used to train a policy network (actor) and a value network (critic) using the Actor-Critic with GAE algorithm in a given environment.

The function has the same structure as `train_reinforce_baseline_discrete` (Exercise 2): for each training episode, it generates an episode with `sample_one_episode` and performs one optimization step per episode for each network. The differences are the following:
1. the values $\hat{V}_{\boldsymbol{\omega}}(s_0), \dots, \hat{V}_{\boldsymbol{\omega}}(s_T)$ of all the states of the episode, *including the last one*, are computed, with $\hat{V}_{\boldsymbol{\omega}}(s_T) = 0$ if $s_T$ is a terminal state;
2. the advantage estimates $\hat{A}_t$ are computed with `compute_gae` (instead of $G_t - \hat{V}_{\boldsymbol{\omega}}(s_t)$), and used in the policy loss $-\sum_{t=0}^{T-1} \hat{A}_t \ln \pi_{\boldsymbol{\theta}}(a_t|s_t)$;

#### Train the agent

Since the networks are updated once per episode, we use the same number of training episodes and the same policy learning rate as in Exercise 2.
The learning rate of the critic is higher than in Exercise 2 (0.1 instead of 0.01): with $\lambda < 1$, the advantage estimates rely on the critic, which is only updated once per episode, so it must learn quickly.

The agent is trained with three values of $\lambda$: $\lambda = 0$ (the advantage estimate of Exercise 3), $\lambda = 0.95$ (the default value in PPO) and $\lambda = 1$ (the advantage estimate of Exercise 2).
You should observe that $\lambda = 0$ learns much more slowly than the two other values: with only one update per episode, the critic remains inaccurate for a long time and the bias of the advantage estimates dominates (in Exercise 3, the same advantage estimate works better because the critic is updated at every time step).
On CartPole, $\lambda = 0.95$ and $\lambda = 1$ give similar results ($\lambda = 0.95$ is often slightly faster): the variance of the Monte Carlo returns remains moderate on this simple task. The variance reduction brought by $\lambda < 1$ is much more useful on harder tasks with longer episodes.
Feel free to try other values of $\lambda$ (*e.g.* 0.5 or 0.9).
